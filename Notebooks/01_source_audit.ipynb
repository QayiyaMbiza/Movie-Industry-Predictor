```python
def numeric_quality_report(
    df,
    column,
    dataset_name
):
    """
    Inspect completeness and zero values
    for a numeric field.
    """

    if column is None:
        print(
            f"\n{dataset_name}: "
            "Column not found."
        )
        return

    numeric = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    total = len(df)

    missing = numeric.isna().sum()
    zero = (numeric == 0).sum()
    positive = (numeric > 0).sum()
    negative = (numeric < 0).sum()

    print("\n")
    print(dataset_name)
    print("-" * 60)

    print(f"Column:              {column}")
    print(f"Total rows:          {total:,}")
    print(f"Missing values:      {missing:,}")
    print(f"Zero values:         {zero:,}")
    print(f"Positive values:     {positive:,}")
    print(f"Negative values:     {negative:,}")

    if total > 0:
        print(
            f"Positive percentage: "
            f"{positive / total * 100:.2f}%"
        )
```

## COUNT MOVIES WITH BOTH BUDGET AND REVENUE

```python
def count_complete_financial_rows(
    df,
    budget_column,
    revenue_column,
    dataset_name
):
    """
    Count rows with valid positive budget and positive revenue.
    """

    if budget_column is None or revenue_column is None:
        print(
            f"{dataset_name}: "
            "Required financial columns not found."
        )
        return 0

    budget = pd.to_numeric(
        df[budget_column],
        errors="coerce"
    )

    revenue = pd.to_numeric(
        df[revenue_column],
        errors="coerce"
    )

    valid = (
        (budget > 0)
        &
        (revenue > 0)
    )

    count = valid.sum()

    percentage = (
        count / len(df) * 100
        if len(df) > 0
        else 0
    )

    print(
        f"{dataset_name}: "
        f"{count:,} movies "
        f"({percentage:.2f}%)"
    )

    return count


print("\n")
print("=" * 80)
print("MOVIES WITH BOTH BUDGET AND REVENUE")
print("=" * 80)

cleaned_complete = count_complete_financial_rows(
    cleaned_movies,
    cleaned_budget_col,
    cleaned_revenue_col,
    "Dataset A - Cleaned Movies"
)

relational_complete = count_complete_financial_rows(
    tmdb_relational,
    relational_budget_col,
    relational_revenue_col,
    "Dataset B - TMDB Relational Movies"
)

recent_complete = count_complete_financial_rows(
    tmdb_recent,
    recent_budget_col,
    recent_revenue_col,
    "Dataset C - TMDB Recent Movies"
)
```

## RELEASE YEAR COVERAGE

```python
def audit_year_range(
        df, 
        dataset_name,
        possible_year_columns,
        possible_date_columns
):
    # Determine the earliest and latest release years.

    year_column = find_existing_column(
        df, 
        possible_year_columns
    )

    date_column = find_existing_column(
        df, 
        possible_date_columns
    )

    years = None

    if year_column is not None:
        years = pd.to_numeric(
            df[year_column], errors="coerce"
        )

    elif date_column is not None:
        dates = pd.to_datetime(
            df[date_column], errors="coerce"
        )

        years = dates.dt.year

    if years is None:

        print(
            f"{dataset_name}: "
            "No valid year or date column found."
        )
        return None, None

    years = years.dropna()

    if years.empty:

        print(
            f"{dataset_name}: "
            "No valid year data found."
        )
        return None, None

    min_year = int(years.min())
    max_year = int(years.max())

    print(
        f"{dataset_name}: "
        f"{min_year} - {max_year}"
    )

    return min_year, max_year

print("\n")
print("="*80)
print("Release Year Coverage")
print("="*80)

cleaned_min_year, cleaned_max_year = (
    audit_year_range(
            cleaned_movies,
            "Dataset A - Cleaned Movies",
            ["release_year", "year"],
            ["release_date", "date"]
    )
)

relational_min_year, relational_max_year = (
    audit_year_range(
        tmdb_relational,
        "Dataset B - TMDB Relational Movies",
        ["release_year", "year"],
        ["release_date", "released"]
    )
)

recent_min_year, recent_max_year = (
    audit_year_range(
        tmdb_recent,
        "Dataset C - TMDB Recent Movies",
        ["release_year", "year"],
        ["release_date", "released"]
    )
)
```

## TITLE DUPLICATES

```python
def audit_title_duplicates(df, dataset_name):
    # Count the number of duplicate movie titles in the dataset.

    title_column = find_existing_column(
        df, ["title", "name", "original_title" ]
    )

    if title_column is None:

        print(
            f"{dataset_name}: "
            "No title column found in the dataset."
        )

        return

    duplicates = (
        df[title_column]
        .astype(str)
        .str.strip()
        .str.lower()
        .duplicated(
            keep=False  
        )
    )

    duplicate_count = duplicates.sum()

    print(
        f"{dataset_name}: "
        f"{duplicate_count:,} rows "
    # belong to duplicated titles
    )

print("\n")
print("="*80)
print("Duplicate Movie Titles")
print("="*80)

audit_title_duplicates(
    tmdb_recent,
    "Dataset C - TMDB Recent Movies"
)
```

## UNIQUE ID AUDIT

```python
def audit_id_column(
    df,
    dataset_name
):
    """
    Inspect possible movie ID columns.
    """

    id_column = find_existing_column(
        df,
        ["id", "movie_id", "tmdb_id", "imdb_id"]
    )

    if id_column is None:
        print(
            f"{dataset_name}: "
            "No movie ID column found."
        )
        return

    total = len(df)

    unique_ids = (
        df[id_column]
        .nunique(
            dropna=True
        )
    )

    missing_ids = (
        df[id_column]
        .isna()
        .sum()
    )

    duplicate_ids = (
        df[id_column]
        .duplicated(
            keep=False
        )
        .sum()
    )

    print("\n")
    print(dataset_name)
    print("-" * 60)

    print(
        f"ID column:       {id_column}"
    )

    print(
        f"Rows:            {total:,}"
    )

    print(
        f"Unique IDs:      {unique_ids:,}"
    )

    print(
        f"Missing IDs:     {missing_ids:,}"
    )

    print(
        f"Duplicate IDs:   {duplicate_ids:,}"
    )


print("\n")
print("=" * 80)
print("MOVIE ID QUALITY")
print("=" * 80)

audit_id_column(
    cleaned_movies,
    "Dataset A - Cleaned Movies"
)

audit_id_column(
    tmdb_relational,
    "Dataset B - TMDB Relational Movies"
)

audit_id_column(
    tmdb_recent,
    "Dataset C - TMDB Recent Movies"
)
```

## CREATE SUMMARY TABLE

```python
def positive_count(
        df,
        column
):
#Count positive numeric values.

    if column is None:
        return None

    values = pd.to_numeric(
        df[column],
        errors='coerce'
    )
    return int(
    (values > 0).sum()
    )

summary = pd.DataFrame(
    [
        {
            "dataset":
                "Dataset A - Cleaned Movies",

            "rows":
                len(cleaned_movies),

            "columns":
            cleaned_movies.shape[1],

            "start_year":
                cleaned_min_year,

            "end_year":
                cleaned_max_year,

            "positive_budget":
                positive_count(
                    cleaned_movies,
                    cleaned_budget_col
                ),

            "positive_revenue":
                positive_count(
                    cleaned_movies,
                    cleaned_revenue_col
                ),

            "budget_and_revenue":
                cleaned_complete
        },
        {
            'dataset':
                'Dataset B - TMDB Relational',

            'rows':
                len(tmdb_relational),

            'columns':
                tmdb_relational.shape[1],

            'start_year':
                relational_min_year,

            'end_year':
                relational_max_year,

            'positive_budget':
                positive_count(
                    tmdb_relational,
                        relational_budget_col
              ),

            'positive_revenue':
                positive_count(
                    tmdb_relational,
                    relational_revenue_col
                ),

            'budget_and_revenue':
                relational_complete
        },
        {
            'dataset':
                'Dataset C - TMDB 2010-2025',

            'rows':
                len(tmdb_recent),

            'columns':
                tmdb_recent.shape[1],

            'start_year':
                recent_min_year,

            'end_year':
                recent_max_year,

            'positive_budget':
                positive_count(
                    tmdb_recent,
                    recent_revenue_col
                ),

            'budget_and_revenue':
                recent_complete
        }
    ]
)

print('\n')
print('='*80)
print('Source Audit Summary')
print('='*80)

print(summary.to_string(index=False))
```
